```python
from pathlib import Path
import pandas as pd
import joblib
```

```python
BASE_DIR = Path.cwd().parent
DATA_DIR = BASE_DIR / "datasets" / "cicids2017" / "processed"
```

```python
def load_processed_data():

    print("Loading processed CICIDS2017 data...")

    files = {
        "X_train_binary": "X_train_binary.csv",
        "X_test_binary": "X_test_binary.csv",
        "y_train_binary": "y_train_binary.csv",
        "y_test_binary": "y_test_binary.csv",

        "X_train_multiclass": "X_train_multiclass.csv",
        "X_test_multiclass": "X_test_multiclass.csv",
        "y_train_multiclass": "y_train_multiclass.csv",
        "y_test_multiclass": "y_test_multiclass.csv"
    }

    data = {}

    for key, filename in files.items():

        path = DATA_DIR / filename

        if not path.exists():
            raise FileNotFoundError(
                f"File not found: {path}"
            )

        data[key] = pd.read_csv(path)

        print(f"{key}: {data[key].shape}")

    print("\nAll datasets loaded successfully!")

    return data
```

```python
def validate_data(data):

    for key, df in data.items():

        print(f"\n{key}")

        print("Shape:", df.shape)
        print("Missing values:", df.isnull().sum().sum())

        if df.isnull().values.any():
            raise ValueError(
                f"Missing values detected in {key}"
            )

    assert (
        data["X_train_binary"].shape[1]
        == data["X_test_binary"].shape[1]
    )

    assert (
        data["X_train_multiclass"].shape[1]
        == data["X_test_multiclass"].shape[1]
    )

    assert (
        len(data["X_train_binary"])
        == len(data["y_train_binary"])
    )

    assert (
        len(data["X_test_binary"])
        == len(data["y_test_binary"])
    )

    assert (
        len(data["X_train_multiclass"])
        == len(data["y_train_multiclass"])
    )

    assert (
        len(data["X_test_multiclass"])
        == len(data["y_test_multiclass"])
    )

    print("\nData validation completed successfully!")
```

```python
if __name__ == "__main__":

    data = load_processed_data()

    validate_data(data)
```

```python
for key in data:

    if key.startswith("X_"):
        data[key] = data[key].to_numpy()

    else:
        data[key] = data[key].iloc[:, 0].to_numpy()
```

```python
data["feature_columns"] = pd.read_csv(
    DATA_DIR / "X_train_binary.csv",
    nrows=0
).columns.tolist()
```

```python
OUTPUT_DIR = BASE_DIR / "datasets" / "cicids2017" / "processed"
OUTPUT_PATH = OUTPUT_DIR / "processed_data.joblib"

joblib.dump(
    data,
    OUTPUT_PATH,
    compress=3
)
```
